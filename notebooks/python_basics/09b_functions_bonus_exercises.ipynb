**🎯 Goal:** Write a function `factorial(n)` that **returns** the factorial of `n` (the product of all positive numbers up to `n`). Assume `n` is a non-negative integer.

**🧪 Example:** `factorial(5)` should return **120** (because 5×4×3×2×1 = 120).

## Exercise 12: Count Vowels in a String

**🎯 Goal:** Create a function `count_vowels(s)` that **returns** the number of vowels (`a, e, i, o, u`) in the string `s`. Treat uppercase and lowercase vowels equally.

**🧪 Example:** `count_vowels("Hello World")` should return **3** (e, o, o).

````python
def count_vowels(s):
  vowels = ('aeiouAEIOU')
  count=0
  for letter in s:
    if letter in vowels:
      count+=1
  return count

````

🔬 Test:

````python
# Expected output: 3
count_vowels("Hello World")
````

**🧪 Example:** `filter_even([1, 2, 3, 4, 5, 6])` should return **[2, 4, 6]**.

## Exercise 29: Palindrome Check

**🎯 Goal:** Write a function `is_palindrome(s)` that **returns** `True` if the string `s` is a palindrome (reads the same forwards and backwards), and `False` otherwise. Ignore case and spaces for simplicity.

**🧪 Example:** `is_palindrome("Racecar")` should return **True**. `is_palindrome("Hello")` should return **False**.

````python
def is_palindrome(s):
  s_clean = s.replace(" ", "").lower()
  return s_clean == s_clean[::-1]
````

🔬 Test:

````python
# Expected output: True
is_palindrome("Racecar")
````

## Exercise 30: Join Words into a Sentence

**🎯 Goal:** Create a function `join_words(words)` that **returns** a single string made by joining a list of words `words` with spaces. For example, `['Python', 'is', 'fun']` becomes "Python is fun".

**🧪 Example:** `join_words(['Python', 'is', 'fun'])` should return **"Python is fun"**.

````python
def join_words(words):
    sentence = ''
    for word in words:
        if sentence:
            sentence += ' '
        sentence += word
    return sentence
````

🔬 Test:

````python
# Expected output: 'I can code'
join_words(['I', 'can', 'code'])
````
