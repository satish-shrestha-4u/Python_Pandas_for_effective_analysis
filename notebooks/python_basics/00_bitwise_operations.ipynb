# Beyond the Basics: Bitwise Operations in Python
### Unlocking the Power of Bits

## What are Bitwise Operations?

- **Working with the Smallest Units:** Bitwise operations directly manipulate individual bits (0s and 1s) of integers.
- **A Low-Level Look:** Instead of operating on the numbers themselves, they work on their binary representations.
- **Why Bother?**
  - Optimizing performance in specific scenarios.
  - Memory-efficient programming.
  - Solving certain algorithmic problems.

## How Do Computers See Numbers?

- **It's All About Binary:** Computers store and process all data, including numbers, in binary format (base 2).
- **Bits and Bytes:** A bit is a single binary digit (0 or 1).
- **Example:** The decimal number 5 is `0101` in binary.

We can see this in Python using the `bin()` function.

```python
for decimal_number in range(20):
    binary_representation = bin(decimal_number)
    print(f"The decimal number  {decimal_number:2}  is  {binary_representation:>7}  in binary.")
```

## Your Toolkit for Bit Manipulation

Python provides six bitwise operators:

- AND (`&`)
- OR (`|`)
- XOR (`^`)
- NOT (`~`)
- Left Shift (`<<`)
- Right Shift (`>>`)

### Bitwise AND: `&`

- **Rule:** The result bit is 1 only if both corresponding bits are 1. Otherwise, it's 0.
- **Analogy:** Imagine two light switches; the light only turns on if both are on.

```python
# Example 1: Basic AND operation
a = 5  # Binary: 0101
b = 3  # Binary: 0011

result = a & b # 0101 & 0011 = 0001

print(f"5 & 3 = {result}")
print(f"Binary: {bin(result)}\n")
```

```python
# Example 2: Checking if a number is even or odd
# An even number always has its last bit as 0. An odd number's last bit is 1.
# So, number & 1 will be 0 if even, and 1 if odd.
num_even = 12 # Binary: 1100
num_odd = 13  # Binary: 1101

print(f"Is {num_even} odd? {(num_even & 1) == 1}")
print(f"Is {num_odd} odd? {(num_odd & 1) == 1}")
```

### Bitwise OR: `|`

- **Rule:** The result bit is 1 if at least one of the corresponding bits is 1. Otherwise, it's 0.
- **Analogy:** The light turns on if either switch is on (or both).

```python
# Example 1: Basic OR operation
a = 5  # Binary: 0101
b = 3  # Binary: 0011

result = a | b # 0101 | 0011 = 0111

print(f"5 | 3 = {result}")
print(f"Binary: {bin(result)}\n")
```

```python
# Example 2: Setting a specific bit (turning a flag 'on')
# Let's say we have a number and we want to ensure the '4s' bit (binary 100) is on.
x = 10 # Binary: 1010
FLAG = 4 # Binary: 0100
x_with_flag = x | FLAG # 1010 | 0100 = 1110

print(f"{x} with 4s bit turned on is {x_with_flag}")
print(f"Binary: {bin(x_with_flag)}")
```

### Bitwise XOR: `^`

- **Rule:** The result bit is 1 if the corresponding bits are different. Otherwise, it's 0.
- **Analogy:** The light turns on if only one switch is on.

```python
# Example 1: Basic XOR operation
a = 5  # Binary: 0101
b = 3  # Binary: 0011

result = a ^ b # 0101 ^ 0011 = 0110

print(f"5 ^ 3 = {result}")
print(f"Binary: {bin(result)}\n")
```

```python
# Example 2: Swapping two numbers without a temporary variable
x = 10
y = 20
print(f"Before swap: x={x}, y={y}")
x = x ^ y
y = x ^ y
x = x ^ y
print(f"After swap:  x={x}, y={y}")
```

### Bitwise NOT: `~`

- **Rule:** Inverts all the bits of the number (0 becomes 1, and 1 becomes 0).
- **Signed Numbers:** Python uses two's complement for negative numbers, so `~x` is equivalent to `-(x + 1)`.

```python
# Example 1
a = 5 # Binary: ...00000101
result = ~a # Result: ...11111010
print(f"~5 = {result}")
print(f"Binary: {bin(result)}\n")

# Example 2
b = -11
result_b = ~b # -(x+1) -> -(-11+1) -> -(-10) -> 10
print(f"~-11 = {result_b}")
```

### Left Shift: `<<`

- **Rule:** Shifts the bits of the number to the left by a specified number of positions. New positions are filled with zeros.
- **Effect:** Equivalent to multiplying the number by 2^n, where n is the number of shifts.

```python
# Example 1: Shift left by 1 position (multiply by 2)
a = 5 # Binary: 0101
result = a << 1 # 0101 -> 1010
print(f"5 << 1 = {result}")
print(f"Binary: {bin(result)}\n")

# Example 2: Shift left by 3 positions (multiply by 2^3 = 8)
b = 4 # Binary: 100
result_b = b << 3 # 100 -> 100000
print(f"4 << 3 = {result_b}")
print(f"4 * 8 = {4 * (2**3)}")
```

### Right Shift: `>>`

- **Rule:** Shifts the bits of the number to the right by a specified number of positions.
- **Effect:** Equivalent to integer division by 2^n, where n is the number of shifts.

```python
# Example 1: Shift right with an even number
a = 20 # Binary: 10100
result = a >> 1 # 10100 -> 1010 (20 / 2 = 10)
print(f"20 >> 1 = {result}")
print(f"Binary: {bin(result)}\n")

# Example 2: Shift right with an odd number (truncates)
b = 21 # Binary: 10101
result_b = b >> 1 # 10101 -> 1010 (21 // 2 = 10)
print(f"21 >> 1 = {result_b}")
print(f"Binary: {bin(result_b)}")
```

## Practical Applications

- **Flags and Permissions:** Efficiently storing and checking multiple true/false states in a single integer.
- **Image Processing:** Manipulating pixel data.
- **Network Protocols:** Parsing and creating data packets.
- **Data Compression/Encryption:** Low-level data manipulation.
- **Optimized Calculations:** Sometimes faster than multiplication/division for powers of two.

### Python Example: Checking a Flag

A common use case is managing a set of permissions. Each permission is a power of two, representing a single bit in a number. We can use `|` to grant permissions and `&` to check them.

```python
# Define permissions as flags (powers of 2)
READ    = 1  # Binary: 001
WRITE   = 2  # Binary: 010
EXECUTE = 4  # Binary: 100

# Grant a user Read and Execute permissions
user_permissions = READ | EXECUTE # 001 | 100 = 101 (which is 5)
print(f"User permissions value: {user_permissions}")
print(f"Binary representation: {bin(user_permissions)}\n")

# Now, let's check if the user has specific permissions

# Check for READ permission
has_read = (user_permissions & READ) == READ
print(f"Has read permission? {has_read}")

# Check for WRITE permission
has_write = (user_permissions & WRITE) == WRITE
print(f"Has write permission? {has_write}")

# Check for EXECUTE permission
has_execute = (user_permissions & EXECUTE) == EXECUTE
print(f"Has execute permission? {has_execute}")
```
