```python
s = "Hello\nWorld\t!"
t = r"Hello\nWorld\t!"
print(s)
print(len(s))
print(repr(s[5]))
print(s[6])
print(s[:5])
print(repr(t[5]))
# Explanation:

# len(s) = 13 (Hello + newline + World + tab + !).
# s[5] is the newline character; repr(s[5]) prints '\n'.
# s[6] is 'W'.
# s[:5] is 'Hello'.
# t is a raw string, so t[5] is a single backslash; repr(t[5]) prints '\'.
```
