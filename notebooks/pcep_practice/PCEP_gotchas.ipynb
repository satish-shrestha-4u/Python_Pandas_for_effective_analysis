### Problem 1.9

**Question:** What will be the output of the following code? Predict the output before running.

````python
print(-7 % 3)
print(7 % -3)
print(-7 % -3)
````

<details>
<summary>Hint</summary>
In Python, the sign of the result of the modulo operation (`%`) is the same as the sign of the *divisor*.
</details>

<details>
<summary>Solution</summary>

```
2
-2
-1
```

</details>

<details>
<summary>Explanation</summary>
Python's modulo operator `%` works such that the result always has the same sign as the divisor (the second operand).

* `-7 % 3`: The divisor is `3` (positive), so the result is positive. `-7 = 3 * (-3) + 2`. The remainder is `2`.
* `7 % -3`: The divisor is `-3` (negative), so the result is negative. `7 = -3 * (-2) + 1`. The remainder is `1`. To match the divisor's sign, `7 = -3 * (-3) - 2`. The remainder is `-2`.
* `-7 % -3`: The divisor is `-3` (negative), so the result is negative. `-7 = -3 * 3 + 2`. The remainder is `2`. To match the divisor's sign, `-7 = -3 * 2 - 1`. The remainder is `-1`.

This behavior can be counter-intuitive compared to some other programming languages where the sign of the result matches the dividend.
</details>

### Problem 2.8

**Question:** What will be the output of the following code? Predict the output before running.

````python
print(5 & 3)
print(5 | 3)
print(5 ^ 3)
print(~5)
print(5 << 1)
print(5 >> 1)
````

<details>
<summary>Hint</summary>
Convert the numbers to their binary representations and perform the operations bit by bit. For `~` (bitwise NOT), remember two's complement representation for negative numbers.
</details>

<details>
<summary>Solution</summary>

```
1
7
6
-6
10
2
```

</details>

<details>
<summary>Explanation</summary>
Let's look at the binary representations (using 8 bits for illustration):
`5` in binary: `0000 0101`
`3` in binary: `0000 0011`

* `5 & 3` (Bitwise AND):
    `0000 0101` (`5`)
    `& 0000 0011` (`3`)
    `----------`
    `0000 0001` (`1`)
    Result is `1`.

* `5 | 3` (Bitwise OR):
    `0000 0101` (`5`)
    `| 0000 0011` (`3`)
    `----------`
    `0000 0111` (`7`)
    Result is `7`.

* `5 ^ 3` (Bitwise XOR):
    `0000 0101` (`5`)
    `^ 0000 0011` (`3`)
    `----------`
    `0000 0110` (`6`)
    Result is `6`.

* `~5` (Bitwise NOT):
    This inverts all bits. For positive `x`, `~x` is equivalent to `-(x + 1)`.
    `0000 0101` (`5`)
    Inverting bits: `1111 1010`. In two's complement, this represents `-6`.
    Result is `-6`.

* `5 << 1` (Left Shift):
    Shifts bits to the left by 1 position, equivalent to multiplying by $2^1$.
    `0000 0101` (`5`) shifted left by 1 becomes `0000 1010` (`10`).
    Result is `10`.

* `5 >> 1` (Right Shift):
    Shifts bits to the right by 1 position, equivalent to integer division by $2^1$.
    `0000 0101` (`5`) shifted right by 1 becomes `0000 0010` (`2`).
    Result is `2`.
</details>

### Problem 2.9

**Question:** What will be the output of the following code? Predict the output before running.

````python
list1 = [1, 2, 3]
list2 = list1
list2.append(4)
print(list1)
print(list2)
````

<details>
<summary>Hint</summary>
When you assign one list to another using `=`, you are creating an alias, not a copy. Both variables point to the same object in memory.
</details>

<details>
<summary>Solution</summary>

```
[1, 2, 3, 4]
[1, 2, 3, 4]
```

</details>

<details>
<summary>Explanation</summary>
In Python, lists are mutable objects. When you write `list2 = list1`, you are not creating a new list. Instead, `list2` becomes a new reference (or "alias") to the *same* list object that `list1` refers to. Therefore, any changes made through `list2` (like `list2.append(4)`) will also be reflected when accessing the list through `list1`, because they are both looking at the exact same data. To create a shallow copy, you would use `list2 = list1.copy()` or `list2 = list1[:]`.
</details>

### Problem 2.10

**Question:** What will be the value of `squares` after the following code executes? Predict the output before running.

````python
numbers = [1, 2, 3, 4]
squares = [x*x for x in numbers if x % 2 == 0]
print(squares)
````

<details>
<summary>Hint</summary>
List comprehensions provide a concise way to create lists. Remember the order: `[expression for item in iterable if condition]`.
</details>

<details>
<summary>Solution</summary>

```
[4, 16]
```

</details>
<details>
<summary>Explanation</summary>
The list comprehension `[x*x for x in numbers if x % 2 == 0]` works as follows:
1.  It iterates through each `x` in the `numbers` list.
2.  For each `x`, it checks if the condition `x % 2 == 0` (is `x` even?) is `True`.
3.  If the condition is `True`, the `expression` (`x*x`) is evaluated, and its result is added to the new `squares` list.

* For `x = 1`: `1 % 2 == 0` is `False`. (Skipped)
* For `x = 2`: `2 % 2 == 0` is `True`. `2*2 = 4` is added.
* For `x = 3`: `3 % 2 == 0` is `False`. (Skipped)
* For `x = 4`: `4 % 2 == 0` is `True`. `4*4 = 16` is added.

The resulting list is `[4, 16]`.
</details>

* **Python 3.7+:** The output order is guaranteed to be `apple`, `banana`, `cherry`.
* **Python 3.6:** The output order is insertion-ordered (an implementation detail, not officially guaranteed until 3.7).
* **Python 3.5 and older:** The output order is *not* guaranteed and could be arbitrary.

</details>

<details>
<summary>Explanation</summary>
This is a common "gotcha" related to Python versioning.

* **Before Python 3.7:** Dictionaries were *unordered* collections. While CPython 3.6 did preserve insertion order as an implementation detail, it was not part of the language specification, meaning other Python implementations (like PyPy or Jython) were not required to maintain this order.

* **Python 3.7 and later:** The insertion-order preservation of dictionary elements became an official part of the Python language specification. This means that when you iterate over a dictionary, the keys (and values, and items) will be returned in the order they were inserted.

It's crucial for exam takers to know this distinction, especially if the exam environment might use an older Python version or if the question implies behavior across versions.
</details>

### Problem 3.8

**Question:** What will be the output of the following code? Predict the output before running.

````python
x = 10

def my_function():
    x = 5
    print(x)

my_function()
print(x)
````

<details>
<summary>Hint</summary>
Python's scope rules dictate that assigning to a variable inside a function creates a new local variable unless `global` is explicitly used.
</details>

<details>
<summary>Solution</summary>

```
5
10
```

</details>

<details>
<summary>Explanation</summary>
When `my_function()` is called, the line `x = 5` creates a *new local variable* named `x` within the scope of `my_function`. This local `x` shadows (hides) the global `x`. So, `print(x)` inside the function prints the local `x` (which is 5).

After the function finishes, the local `x` is destroyed. The `print(x)` outside the function refers to the global `x`, which was never modified and still holds its initial value of 10. If you wanted to modify the global `x` inside the function, you would need to use the `global` keyword:

```python
x = 10

def my_function():
    global x
    x = 5
    print(x)

my_function()
print(x)
```
</details>

### Problem 3.9

**Question:** What will be the output of the following code? Predict the output before running.

````python
def add_item(item, my_list=[]):
    my_list.append(item)
    return my_list

print(add_item(1))
print(add_item(2))
print(add_item(3, [4, 5]))
print(add_item(6))
````

<details>
<summary>Hint</summary>
Default arguments are evaluated *only once* when the function is defined, not every time the function is called. This can lead to unexpected behavior with mutable default arguments.
</details>

<details>
<summary>Solution</summary>

```
[1]
[1, 2]
[4, 5, 3]
[1, 2, 6]
```

</details>

<details>
<summary>Explanation</summary>
This is a classic Python "gotcha." When `my_list=[]` is defined as a default argument, the empty list `[]` is created *once* when the `add_item` function is first defined (i.e., when the Python interpreter reads the `def` statement).

* **`print(add_item(1))`**: `my_list` is the default `[]`. `1` is appended. Returns `[1]`.
* **`print(add_item(2))`**: `my_list` is *still* the same list object that was created initially. It now contains `[1]`. `2` is appended to *that same list*. Returns `[1, 2]`.
* **`print(add_item(3, [4, 5]))`**: Here, a *new* list `[4, 5]` is explicitly passed as `my_list`. `3` is appended to *this new list*. Returns `[4, 5, 3]`. The original default list is unaffected.
* **`print(add_item(6))`**: Again, `my_list` defaults to the *original* list object, which now contains `[1, 2]`. `6` is appended to it. Returns `[1, 2, 6]`.

To avoid this, a common pattern is to use `None` as the default and initialize the list inside the function:

```python
def add_item_corrected(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list

print(add_item_corrected(1))
print(add_item_corrected(2))
print(add_item_corrected(3, [4, 5]))
print(add_item_corrected(6))
```
</details>

### Problem 3.14

**Question:** What will be the output of the following code? Predict the output before running.

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(list1 == list2)
print(list1 is list2)
print(list1 == list3)
print(list1 is list3)

a = 1
b = 1
print(a is b)

x = 300
y = 300
print(x is y)
````

<details>
<summary>Hint</summary>
`==` checks for *equality* (do they have the same value?). `is` checks for *identity* (do they refer to the exact same object in memory?). Python optimizes small integers.
</details>

<details>
<summary>Solution</summary>

```
True
False
True
True
True
False
```

</details>

<details>
<summary>Explanation</summary>

* `list1 == list2`: `True`. Both lists contain the same elements in the same order, so their *values* are equal.
  
* `list1 is list2`: `False`. `list1` and `list2` are two separate list objects in memory, even though they have identical content. The `is` operator checks if they are the *exact same object*.
  
* `list1 == list3`: `True`. `list3` is an alias for `list1`, so they refer to the same object, and thus their values are equal.
  
* `list1 is list3`: `True`. `list3` was assigned `list1`, meaning both variables point to the exact same list object in memory.

* `a = 1`, `b = 1`, `print(a is b)`: `True`. This is due to a CPython optimization where small integers (typically from -5 to 256) are 'interned' or 'cached'. This means that when you create multiple variables with these small integer values, Python often points them all to the *same* immutable object in memory for efficiency.
  
* `x = 300`, `y = 300`, `print(x is y)`: `False`. For larger integers, Python typically creates distinct objects even if their values are the same. Therefore, `x` and `y` refer to different objects in memory.

Always use `==` for value comparison and `is` for identity comparison.
</details>
